### Preguntas
    ¿Qué recibe? Un número (entero o flotante) representado por el parámetro valor.

    ¿Qué calcula? La multiplicación del valor por 2 (valor * 2).

    ¿Qué retorna? El resultado numérico del cálculo realizado.

    ¿Dónde queda almacenado el resultado? En el punto desde donde se invocó la función; se puede asignar a una variable o pasar directamente a otra función como print().

## Reto 5

```python
def clasificar_valor(valor):
    if valor <= 0 or valor > 100:
        return("ERROR")
    elif valor < 70:
        return("BAJO")
    elif valor < 95:
        return("NORMAL")
    else:
        return("ALTO")

print(clasificar_valor(-2))
print(clasificar_valor(0))
print(clasificar_valor(69))
print(clasificar_valor(70))
print(clasificar_valor(94))
print(clasificar_valor(95))
print(clasificar_valor(100))
```

## Reto 6

```python
def contar_registros(valores):
    i = 0  
    for valor in valores:
        i += 1
    return i


print(contar_registros([22, 55, 72, 38, 61]))  
```

## Reto 7

```python
def calcular_suma(valores):
    i = 0  
    for valor in valores:
        i = valor + i
    return i

print(calcular_suma([10, 20, 30, 40]))
```

## Reto 8

```python
def clasificar_valor(valor):
    BAJO = 0
    NORMAL = 0
    ALTO = 0
    if valor <= 0 or valor > 100:
        return("ERROR")
    elif valor < 70:
        BAJO += 1
        return("BAJO")
    elif valor < 95:
        NORMAL += 1
        return("NORMAL")
    else:
        ALTO += 1
        return("ALTO")
    
valores = [20, 55, 80, 40, 70]
print(clasificar_valor(valores))
print("Total bajo: ", BAJO)
print("Total normal: ", NORMAL)
print("Total alto: ", ALTO)



```
