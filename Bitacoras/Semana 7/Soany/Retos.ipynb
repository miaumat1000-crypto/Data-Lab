## Reto 1 — Primera función.

```python
def mostrar_mensaje():
    print("Procesando Dataset de Trabajo")

# Invocación
mostrar_mensaje()
```

1. La definición es def mostrar_mensaje(): ...
2. La invocación es mostrar_mensaje().
3. La responsabilidad de la función es mostrar el mensaje inicial en pantalla.

## Reto 2

```python
def saludar(nombre):
    print(f"Hola {nombre}")

saludar("Ana")
saludar("Carlos")
saludar("María")
```

1. El parámetro es nombre (en la definición).   
2. El argumento son los valores "Ana", "Carlos", "María" (en la llamada).   
3. La ventaja del parámetro es que permite reutilizar la misma función con diferentes datos sin duplicar código. 

## Retos 3 y 4

```python
def duplicar(valor):
    return valor * 2

print(duplicar(5))   # Salida: 10
print(duplicar(0))   # Salida: 0
print(duplicar(10))  # Salida: 20
```

print() solo muestra información visualmente en la pantalla, mientras que return entrega un valor procesado para que otro bloque del programa pueda utilizarlo.
