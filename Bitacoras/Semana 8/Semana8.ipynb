## Reto 1:


```python
# 1. Crea la lista valores con los cinco datos indicados.
valores = [22.5, 55.1, 72.3, 38.7, 61.4]

# 2. Imprime la lista.
print("Lista de valores:", valores)

# 3. Imprime el primer elemento, el último elemento y la cantidad.
print("Primer elemento:", valores[0])
print("Último elemento:", valores[-1]) # O usando valores[len(valores) - 1]
print("Cantidad de elementos:", len(valores)) 
```

## Reto 2:

```python
valores = [22.5, 55.1, 72.3]

# Muestra el primer elemento.
print(valores[0])
# Muestra el segundo elemento.
print(valores[1])
# Muestra el tercer elemento.
print(valores[2])

# ¿Cuál es el primer índice y por qué?
# El primer índice es 0 porque en Python y otros lenguajes de programación las estructuras están basadas en cero, el índice es el "desplazamiento" o distancia desde el inicio de la lista.
```

## Reto 3:

```python
valores = [10, 20, 30, 40, 50]

# Calcula el último índice usando len().
ultimo_indice = len(valores) - 1

# Imprime el último elemento.
print("El último elemento es:", valores[ultimo_indice])
```
