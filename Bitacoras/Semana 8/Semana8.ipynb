## Reto 9:

````python
def calcular_suma(valores):
    suma = 0 # Inicializa el acumulador
    for valor in valores:
        suma += valor # Recorre la lista y acumula
    return suma # Retorna la suma

prueba = [10, 20, 30, 40]
print(calcular_suma(prueba))
````

## Reto 10:

````python
def contar_elementos(valores):
    contador = 0 # Inicializa un contador
    for valor in valores:
        contador += 1 # Aumenta el contador por cada elemento
    return contador # Retorna el contador

print(contar_elementos([10, 20, 30, 40, 50]))

# Diferencia con len(): 
# len() es una función nativa optimizada de Python que consulta directamente el metadato 
# de longitud guardado en la estructura interna de la lista de forma instantánea. 
# La función manual recorre elemento por elemento sumando uno a uno mediante código explícito.
````

## Reto 11:

````python
def clasificar_valor(valor):
    if valor >= 60:
        return "ALTO"
    elif valor >= 35:
        return "NORMAL"
    return "BAJO"

def procesar_valores(valores):
    for valor in valores:
        clasificacion = clasificar_valor(valor)
        print(f"Valor: {valor} -> Clasificación: {clasificacion}")

procesar_valores([22.5, 55.1, 72.3, 38.7, 61.4])
````

## Reto 12:

````python
def generar_resumen(valores):
    altos = 0
    normales = 0
    bajos = 0

    for valor in valores:
        cat = clasificar_valor(valor)
        if cat == "ALTO":
            altos += 1
        elif cat == "NORMAL":
            normales += 1
        elif cat == "BAJO":
            bajos += 1

    return altos, normales, bajos

print(generar_resumen([22.5, 55.1, 72.3, 38.7, 61.4]))
````

## Reto 13:

````python
def generar_resumen_completo(valores):
    altos = 0
    normales = 0
    bajos = 0
    suma = 0
    existe_alto = False # Bandera inicializada en False

    for valor in valores:
        suma += valor # Acumula la suma
        cat = clasificar_valor(valor)
        
        if cat == "ALTO":
            altos += 1
            existe_alto = True # Se activa la bandera si encuentra un valor ALTO
        elif cat == "NORMAL":
            normales += 1
        elif cat == "BAJO":
            bajos += 1

    return altos, normales, bajos, suma, existe_alto

print(generar_resumen_completo([22.5, 55.1, 72.3, 38.7, 61.4]))
````

````python
def calcular_promedio(valores):
    # Controla la lista vacía para evitar división por cero
    if len(valores) == 0:
        return 0.0 # O un mensaje/valor por defecto definido
    
    suma = sum(valores) # O usando tu función calcular_suma(valores)
    cantidad = len(valores)
    
    return suma / cantidad

print(calcular_promedio([10, 20, 30, 40]))
print(calcular_promedio([]))
````

## Reto 21:

1. **Cómo se almacenan los valores:**
   De manera secuencial en una estructura de datos tipo **lista** de Python (ej. `valores = [22.5, 55.1, 72.3]`), donde cada elemento ocupa una posición que comienza desde el índice `0` hasta `len(valores) - 1`. Pueden capturarse de forma dinámica mediante el método `append()` a partir de entradas del usuario.

2. **Qué función recibe la lista:**
   Las funciones principales del sistema, tales como `calcular_suma()`, `contar_elementos()`, `calcular_promedio()`, `procesar_valores()`, `generar_resumen()` y `generar_resumen_completo()`, reciben la lista de valores directamente como **parámetro** (argumento de entrada). Esto permite modularizar el código y procesar conjuntos de datos de cualquier tamaño sin depender de variables globales.

3. **Cómo se recorre y clasifica:**
   La lista se recorre iterando a través de sus elementos utilizando bucles (como `for valor in valores` para recorrer por valor, o `for i in range(len(valores))` para recorrer por índice). Durante el recorrido, cada elemento se evalúa y se pasa a la función reutilizable `clasificar_valor(valor)`, la cual compara el número con umbrales predefinidos para asignarle una categoría de texto (`"ALTO"`, `"NORMAL"` o `"BAJO"`).

4. **Cómo se genera el resumen:**
   El resumen se genera mediante la función `generar_resumen_completo()`, la cual inicializa previamente contadores independientes por categoría (`altos`, `normales`, `bajos`), un acumulador para la suma total y una variable lógica o bandera (`existe_alto`) en `False`. A medida que el ciclo `for` recorre la lista, acumula la suma, incrementa el contador respectivo según la clasificación obtenida y conmuta la bandera a `True` en cuanto detecta al menos un valor de categoría `"ALTO"`. Finalmente, retorna todos estos agregados consolidados para ser impresos por una función de visualización.

5. **Qué ocurre si la lista está vacía:**
   Se implementa una **condición de control preventiva** (ej. `if len(valores) == 0:`) al inicio de los procesos. Si la lista no contiene elementos, el programa evita la ejecución de bucles innecesarios y previene errores críticos de ejecución como la **división entre cero** al calcular promedios. En su lugar, el sistema intercepta el caso notificando al usuario con un mensaje de advertencia o retornando un estado vacío controlado.

## Respuestas de preguntas finales:

¿Qué problema resuelve una lista?
Permite almacenar múltiples valores en una sola variable organizada secuencialmente.

¿Qué diferencia hay entre índice y valor?
El índice es la posición numérica (dirección), y el valor es el dato guardado en esa posición.

¿Qué diferencia hay entre len() y el último índice?
len() cuenta la cantidad total de elementos (empieza en 1), mientras que el último índice es len() - 1 (porque se cuenta desde el 0).

¿Cuándo conviene recorrer por índice?
Cuando necesitas modificar los elementos originales dentro de la lista.

¿Por qué pasar una lista como parámetro?
Para modularizar el código, permitiendo que una función procese cualquier conjunto dinámico de datos sin depender de variables globales.

¿Por qué reutilizar clasificar_valor()?
Para evitar duplicar código (principio DRY - Don't Repeat Yourself) y mantener la coherencia lógica en todo el proyecto DataLab.

¿Qué relación existe entre listas, contadores y acumuladores?
Las listas proveen los datos, los contadores miden cuántas veces ocurre algo (frecuencias), y los acumuladores suman magnitudes numéricas (totales).

¿Qué nueva capacidad obtiene DataLab v0.8 frente a DataLab v0.7?
La habilidad de procesar conjuntos ilimitados de datos de forma dinámica en lugar de trabajar solo con valores individuales o aislados.
