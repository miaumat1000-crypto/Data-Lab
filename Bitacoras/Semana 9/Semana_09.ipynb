## Reto 1

```python
registro_lista = ["R001", "temperatura", 72.3, "C"]

print("Posición 0:", registro_lista[0])
print("Posición 1:", registro_lista[1])
print("Posición 2:", registro_lista[2])
print("Posición 3:", registro_lista[3])
```

Depender solo de posiciones numéricas dificulta la lectura del código y aumenta el riesgo de errores si los elementos cambian de orden.

## Reto 2

```python
registro_tupla = ("R001", "temperatura", 72.3, "C")

print("ID:", registro_tupla[0])
print("Tipo:", registro_tupla[1])
print("Valor:", registro_tupla[2])
print("Unidad:", registro_tupla[3])
```

 * A diferencia de una lista, una tupla es inmutable; una vez creada no se puede alterar su tamaño ni modificar sus elementos individuales por reasignación de índice.

## Reto 3

```python
registro = ("R001", "temperatura", 72.3, "C")

try:
    registro[2] = 80
except TypeError as error:
    print("Error detectado:", error)
```

 * Se genera un TypeError ("'tuple' object does not support item assignment").
 * Las tuplas son útiles cuando se requiere garantizar que los datos permanezcan protegidos e inalterados durante toda la ejecución del programa.

## Reto 4

 * Colección que puede modificarse: Lista []
 * Colección ordenada que no se modifica: Tupla ()
 * Acceso por índice: Tanto las listas como las tuplas permiten acceso por índice [i]
 * ¿La estructura se elige solo porque es fácil de escribir? No, la elección depende del
 * propósito del dato: si requiere mutabilidad (lista) o integridad inmutable (tupla).

## Reto 15

```python
def clasificar_valor(valor):
    if valor >= 60:
        return "ALTO"
    elif valor >= 35:
        return "NORMAL"
    return "BAJO"

registros = [
    {"id": "R001", "tipo": "temperatura", "valor": 22.5, "unidad": "C"},
    {"id": "R002", "tipo": "temperatura", "valor": 55.1, "unidad": "C"},
    {"id": "R003", "tipo": "temperatura", "valor": 72.3, "unidad": "C"}
]

for registro in registros:
    clasificacion = clasificar_valor(registro["valor"])
    print(f"ID: {registro['id']} | Valor: {registro['valor']} -> Clasificación: {clasificacion}")
```

## Reto 16

```python
for registro in registros:
    clasificacion = clasificar_valor(registro["valor"])
    print(f"ID: {registro['id']} | Tipo: {registro['tipo']} | Valor: {registro['valor']} {registro['unidad']} | Clasificación: {clasificacion}")

# Explicación: Conservar el contexto garantiza que no se pierda la identidad (ID), el tipo
# de magnitud medida ni la unidad del dato procesado a lo largo del análisis.
```

 * Conservar el contexto garantiza que no se pierda la identidad (ID), el tipo de magnitud medida ni la unidad del dato procesado a lo largo del análisis.

## Reto 17

```python
def procesar_registros(registros):
    for registro in registros:
        clasificacion = clasificar_valor(registro["valor"])
        print(f"[{registro['id']}] {registro['tipo']}: {registro['valor']} {registro['unidad']} ({clasificacion})")

procesar_registros(registros)

# Explicación: La lista se pasa como parámetro para desacoplar la función de variables globales,
# haciéndola modular y capaz de procesar cualquier conjunto de registros estructurados.
```

 * La lista se pasa como parámetro para desacoplar la función de variables globales, haciéndola modular y capaz de procesar cualquier conjunto de registros estructurados.

## Reto 18

```python
def generar_resumen(registros):
    altos = 0
    normales = 0
    bajos = 0
    suma = 0.0
    existe_alto = False

    for registro in registros:
        valor = registro["valor"]
        suma += valor  # Acumulador de magnitudes
        
        clasificacion = clasificar_valor(valor)
        
        if clasificacion == "ALTO":
            altos += 1
            existe_alto = True  # Bandera
        elif clasificacion == "NORMAL":
            normales += 1
        elif clasificacion == "BAJO":
            bajos += 1

    return altos, normales, bajos, suma, existe_alto

# Prueba del resumen
altos, normales, bajos, suma, existe_alto = generar_resumen(registros)
print(f"Resumen -> Altos: {altos}, Normales: {normales}, Bajos: {bajos}, Suma total: {suma}, ¿Existe ALTO?: {existe_alto}")
```
